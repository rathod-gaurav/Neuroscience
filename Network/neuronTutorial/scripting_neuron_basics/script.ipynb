### Step 5 : insert a stimulus

```python
iclamp = n.IClamp(soma(0.5))
```

```python
print([item for item in dir(iclamp) if not item.startswith('__')])
```

```python
iclamp.delay = 2 #ms - the time the current clamp switches on after the simulation starts
iclamp.dur = 0.1 #ms - how long the current clamp stays on
iclamp.amp = 0.9 #nA - the amplitude of the current clamp
```

```python
soma.psection()
```

### Step 6 : set up recording variables

```python
v = n.Vector().record(soma(0.5)._ref_v) #membrane potential vector
t = n.Vector().record(n._ref_t) #time stamp vector
```

### Step 7 : run the simulation

```python
n.load_file("stdrun.hoc")
```

```python
n.finitialize(-65*mV)
```

```python
n.continuerun(40*ms)
```

### Step 8 : plot the results

```python
from bokeh.io import output_notebook
import bokeh.plotting as plt
import numpy as np
output_notebook()
```

```python
f = plt.figure(x_axis_label="t (ms)", y_axis_label="v (mV)", title="membrane potential vs time")
f.line(np.array(t), np.array(v), line_width = 1)
plt.show(f)
```
