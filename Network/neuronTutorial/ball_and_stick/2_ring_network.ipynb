### class definitions - generic call and BallAndStick derived from it

```python
#the generic cell class
class Cell:
    def __init__(self, gid, x, y, z, theta):
        self._gid = gid
        self._setup_morphology()
        self.all = self.soma.wholetree()
        self._setup_biophysics()

        self.x = self.y = self.z = 0.0 #position of point in 3D space
        n.define_shape() #creates 3d points for sections, as our sections donot have any yet
        self._set_position(x,y,z)
        self._rotate_z(theta)

    def __repr__(self):
        return "{}[{}]".format(self.name, self._gid)

    def _set_position(self, x, y, z):
        for sec in self.all:
            for i in range(sec.n3d()): #loop over all the points
                sec.pt3dchange(i, x - self.x + sec.x3d(i), y - self.y + sec.y3d(i), z - self.z + sec.z3d(i), sec.diam3d(i))

        self.x, self.y, self.z = x, y, z

    def _rotate_z(self,theta): #rotate the cell about z-axis
        for sec in self.all:
            for i in range(sec.n3d()):
                x = sec.x3d(i)
                y = sec.y3d(i)
                c = n.cos(theta)
                s = n.sin(theta)
                xprime = x*c - y*s
                yprime = x*s + y*c
                sec.pt3dchange(i, xprime, yprime, sec.z3d(i), sec.diam3d(i))
```

```python
#the ball and stick class - derived from Cell class above
class BallAndStick(Cell):
    name = "BallAndStick"

    def _setup_morphology(self):
        self.soma = n.Section("soma", self)
        self.dend = n.Section("dend", self)
        self.dend.connect(self.soma)

        self.soma.L = self.soma.diam = 12.6157
        self.dend.L = 200
        self.dend.diam = 1

    def _setup_biophysics(self):
        for sec in self.all:
            sec.Ra = 100
            sec.cm = 1
        self.soma.insert(n.hh)
        for seg in self.soma:
            seg.hh.gnabar = 0.12
            seg.hh.gkbar = 0.036
            seg.hh.gl = 0.0003
            seg.hh.el = -54.3
        self.dend.insert(n.pas)
        for seg in self.dend:
            seg.pas.g = 0.001
            seg.pas.e = -65
```
